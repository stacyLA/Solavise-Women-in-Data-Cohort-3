```python
# Task 2.1: Calculate total stock value (price multiplied by quantity)
def total_value(price, quantity):
    return price * quantity

# Test with Notebook (price 1500, quantity 40)
print(total_value(1500, 40))
```

    It returns the total stock value by multiplying price by quantity.


```python
# Task 2.2: Calculate discounted price with a default 10% discount
def apply_discount(price, percent=10):
    return price * (1 - percent / 100)

# Test using default (10%) and 25%
print(apply_discount(1500))
print(apply_discount(1500, 25))
```

    it return the price after deducting the specified percentage discount.


```python
# Task 2.3: Return stock availability status based on quantity
def stock_status(quantity):
    if quantity == 0:
        return "Out of stock"
    elif quantity < 10:
        return "Low stock"
    else:
        return "In stock"

# Test with 0, 5, and 40
print(stock_status(0))
print(stock_status(5))
print(stock_status(40))
```

    it return 'Out of stock' if 0, 'Low stock' if below 10, and 'In stock' otherwise.


```python
# Task 2.4: Return list of product names in a given category
def products_in_category(inventory, category):
    return [item["product"] for item in inventory if item["category"] == category]

# Test with "Electronics"
print(products_in_category(inventory, "Electronics"))
```

    it returns a list of product names that belong to the specified category.


```python
# Task 3.1: Print shape and data type of daily_sales
print("Shape:", daily_sales.shape)
print("Data type:", daily_sales.dtype)
```

```python
# Task 3.2: Compute total, mean, max, and standard deviation of daily sales
total_sales = np.sum(daily_sales)
mean_sales = round(np.mean(daily_sales), 2)
highest_sales = np.max(daily_sales)
std_sales = round(np.std(daily_sales), 2)

print("Total:", total_sales)
print("Mean:", mean_sales)
print("Highest:", highest_sales)
print("Standard Deviation:", std_sales)
```

```python
# Task 3.3: Reshape daily_sales into 2 rows (weeks) and 7 columns (days)
weekly_sales = daily_sales.reshape(2, 7)

print(weekly_sales)
print("New shape:", weekly_sales.shape)
```

```python
# Task 3.4: Calculate median and compare it with the mean
median_sales = np.median(daily_sales)

print("Median:", median_sales)
print("Mean:", round(np.mean(daily_sales), 2))
```


The **median (252.5)** describes a typical day much better than the **mean (310.71)**. Across the 14 days, 13 of the days fall tightly between 220 and 320, while Day 13 has an extreme outlier of **980**. Because the mean adds all values together, that single unusually high spike pulls the average upward above almost every normal day, whereas the median finds the middle value and is not distorted by outliers.
